# ZCU216 Board Self Tests
----

## Aims
* Enumerate notebooks and tests to verify the ZCU216 board works as expected.
* Give ZCU216 users a set of tests to report any issues to the PYNQ team.
* Also visit the common page for additional self-tests.

----
